```python
# 시리즈 원소에 함수 매핑
# Series객체.apply(매핑함수)

import seaborn as sns
import pandas as pd

titanic = sns.load_dataset('titanic')
df = titanic.loc[:, ['age', 'fare']]
df.head()
```

```python
def add_10(n):
    return n + 10

def add_two_obj(a, b):
    return a + b

print(add_10(10))
print(add_two_obj(10, 10))
```

```python
df['age'].head()
```

```python
sr1 = df['age'].apply(add_10)
sr1.head()
```

```python
imsi = lambda n: n + 10
imsi(10)
```

```python
sr2 = df['age'].apply(lambda n: n + 10)
sr2.head()
```

```python
sr3 = df['age'].apply(add_two_obj, b=10)
sr3.head()
```

```python
sr4 = df['age'].apply(lambda a, b: a + b, b=10)
sr4.head()
```

```python
# 데이터프레임의 각 결에 함수 매핑 p.358
df

def calculate_state(col):
    max_val = col.max()
    min_val = col.min()
    mean_val = col.mean()
    median_val = col.median()

    return pd.Series(
        [max_val, min_val, mean_val, median_val],
        index=['Max', 'Min', 'Mean', 'Median']
    )

result_df = df.apply(calculate_state, axis=0)
result_df
```

```python
result_sr = df.apply(lambda x: x.max() - x.min())
result_sr
```

```python
df.head()
```

```python
def calculate_diff_avg(row):
    diff = row.max() - row.min()
    avg = row.mean()

    return pd.Series([diff, avg], index=['차이', '평균'])

result_df2 = df.apply(calculate_diff_avg, axis=1)
result_df2
```

```python
titanic = sns.load_dataset('titanic')
df1 = titanic.loc[:, ['age', 'fare']]
df2 = titanic.loc[:, ['embark_town', 'embarked']]

def missing_value(df):
    return df.isnull()

result_df = df1.pipe(missing_value)
result_df
```

```python
def missing_count(df):
    return missing_value(df).sum()

df1.pipe(missing_count)

# df1.isnull().sum()
```

```python
def extract_initial(df):
    df['town_initial'] = df['embark_town'].str[0]
    return df

def verify_initial(df):
    df['verified'] = df['embarked'] == df['town_initial']
    return df

imsi = extract_initial(df2)
verify_initial(imsi)
```

```python
df2.pipe(extract_initial).pipe(verify_initial)
```
