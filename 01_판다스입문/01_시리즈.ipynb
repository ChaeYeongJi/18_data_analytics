```python
import pandas as pd

list_data = [
    '2019-01-02', 3.14, 'ABC', 100, True
]

sr = pd.Series(list_data)

print(sr)
```

```python
print(sr.index)
print(type(sr.values))
print(sr.dtype)
print(len(sr))
print(sr.shape) # * 5행
print(sr.ndim)  # 1차원
```

```python
# 시리즈 원소 선택
import pandas as pd

tup_data = ['영인', '2010-05-01', '여', True]
sr = pd.Series(tup_data, index=['이름', '생년월일', '성별', '학생여부'])
print(sr)
```

```python
# 시리즈 원소 선택
print(sr['이름'])
print(sr['성별'])

# print(sr.iloc[0])
# print(sr[0]) => 지원 안 함
```

```python
# print(sr[[1, 2]]) => 더이상 지원 안 함
print(sr[['생년월일', '성별']])
```

```python
# print(sr[1:2]) => 되기는 함
print(sr['생년월일':'성별'])
print('-'*5)
print(sr[['생년월일', '성별']])
print('='*10)
print(sr['생년월일':'학생여부'])
print('-'*5)
print(sr[['생년월일','학생여부']])
print('-'*5)
print(sr.iloc[1:3])
```
