```python
import pandas as pd

data = [
    [15, '남', '덕영중'],
    [17, '여', '수리중']
]

df = pd.DataFrame(data, index=['준서', '예은'], columns=['나이', '성별', '학교'])
print(df)
```

```python
df.rename(columns={'나이': '연령', '성별': '남녀', '학교': '소속'})
```

```python
print(df)
# rename만으로는 update가 안 되었음을 확인.

# 그래서 재할당
df = df.rename(columns={'나이': '연령', '성별': '남녀', '학교': '소속'})
print(df)
```

```python
df = df.rename(index={'준서': '학생1', '예은': '학생2'})
print(df)
```

```python
# 행/열 삭제
# 행 삭제: DataFrame객체.drop(행 인덱스 또는 배열, axis=0 또는 axis='index')    => 여기서 기본은 axis=0으로 가로 방향. 나중에 다른 거 나옴
# 열 삭제: DataFrame객체.drop(열 이름 또는 배열, axis=1 또는 axis='columns')

import pandas as pd

exam_data = {
    '수학': [90, 80, 70],
    '영어': [98, 89, 95],
    '음악': [85, 95, 100],
    '체육': [100, 90, 90],
}

df = pd.DataFrame(exam_data, index=['서준', '우현', '인아'])
print(df)
```

```python
df2 = df.copy()

# 재할당 필요
df2 = df2.drop('우현')
print(df2)
```

```python
df3 = df.copy()

df3 = df3.drop(['우현', '인아'], axis=0)
print(df3)
```

```python
df4 = df.copy()

df4 = df4.drop(['우현', '인아'], axis='index')
print(df4)
```

```python
df5 = df.copy()

df5.drop(index=['우현'])
```

```python
# 열 삭제
df2 = df.copy()
df2 = df2.drop('수학', axis=1)
print(df2)
```

```python
df3 = df.copy()
df3 = df3.drop(['영어', '음악'], axis=1)
print(df3)
```

```python
df4 = df.copy()
df4 = df4.drop(['영어', '체육'], axis='columns')
print(df4)
```

```python
df5 = df.copy()
df5 = df5.drop(columns=['수학'])
print(df5)
```
